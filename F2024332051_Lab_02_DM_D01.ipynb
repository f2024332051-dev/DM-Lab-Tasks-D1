Write a Python class named 'Rectangle' with attributes length and width. Include
methods to calculate area and perimeter.

```python
class rectangle:
  def __init__(self,l,w):
    self.length = l
    self.width = w
  def area(self):
    return self.length*self.width
  def perimeter(self):
    return 2*(self.length+self.width)

r1 = rectangle(10,10)
print(r1.area())
print(r1.perimeter())
```

Write a Python function that takes a number as input and returns True if it is a prime
number, otherwise False.

```python
def is_prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True

    if n % 2 == 0 or n % 3 == 0:
        return False

    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6

    return True

print(is_prime(13))
```

Build a simple Python program that uses scikit-learn to train a linear regression model
on a small dataset and predicts output for new input.

```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [2, 4, 6, 8, 10]

model = LinearRegression()
model.fit(X, y)

result = model.predict([[6]])

print("Predicted output:", result[0])
```
